# EasyVitessce Tutorial: Non-spatial Example

## Import EasyVitessce

This import statement enables interactive plots by default.
Refer to the [EasyVitessce documentation](https://vitessce.github.io/easy_vitessce/) for how to disable interactive plotting or configure other behaviors.

```python
import easy_vitessce as ev
```

## Import dependencies

```python
import scanpy as sc
```

Here, we load an example dataset as an AnnData object using the [Scanpy Datasets API](https://scanpy.readthedocs.io/en/stable/api/datasets.html).

```python
adata = sc.datasets.pbmc68k_reduced()
adata
```

## Plot the data

```python
sc.pl.pca(adata, color="louvain")
```

```python
sc.pl.dotplot(adata, var_names=["C1QA", "PSAP", "CD79A", "CD79B", "CST3", "LYZ"], groupby="bulk_labels")
```

```python
ev.configure_plots(disable_plots=["pca", "dotplot"])
```
