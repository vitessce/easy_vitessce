```python
import zarr
from zarr.storage import ObjectStore
from obstore.store import S3Store

s3_store = S3Store('idr', skip_signature=True, endpoint="https://livingobjects.ebi.ac.uk")
store = zarr.storage.ObjectStore(store=s3_store, read_only=True)
zarr_group = zarr.open_group(store=store, mode='r', path="zarr/v0.4/idr0062A/6001240.zarr")
```

```python
list(zarr_group.keys())
```

```python
dict(zarr_group.attrs)
```

```python
img_arr = zarr_group['/1']
```

```python
from spatialdata import SpatialData
from spatialdata.models import Image3DModel

image = Image3DModel.parse(img_arr[()], dims=("c","z","y","x"), c_coords=["LaminB1", "Dapi"])
```

```python
sdata = SpatialData(images={"image": image})
sdata
```

```python
vw = (
    sdata
        .pl.render_images("image")
        .pl.show()
)
```
